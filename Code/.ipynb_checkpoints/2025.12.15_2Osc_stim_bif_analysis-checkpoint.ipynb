# Two Nonlinear Two-Population Oscillators with Perturbation
## Model

The two populations' temporal dynamics (change of state) is described by two first-order nonlinear differential equations. Here we have two coupled units:

$$ \frac{dEx_1}{dt}      = (h_{ex}      - Ex_1    - c_2*sigmoid(In_1) + c_{EE}*sigmoid((Ex_1 + 
frac_E*Ex_2))  + pert)*\tau_{ex}$$
$$ \frac{dIn_1}{dt} \,\, = (h_{in} \,\, - In_1 \, - c_4*sigmoid(In_1) + c_{EI}*sigmoid((Ex_1 + 
frac_I*Ex_2)))*\tau_{in}$$

$$ \frac{dEx_2}{dt}      = (h_{ex}      - Ex_2    - c_2*sigmoid(In_2) + c_{EE}*sigmoid((Ex_2 + 
frac_E*Ex_1)))*\tau_{ex}$$
$$ \frac{dIn_2}{dt} \,\, = (h_{in} \,\, - In_2 \, - c_4*sigmoid(In_2) + c_{EI}*sigmoid((Ex_2 + 
frac_I*Ex_1)))*\tau_{in}$$

$Ex_i$ and $In_i$ are variables that change with time $t$ and represent the excitatory and inhibitory population, respectively.

$h_{ex}$ and $h_{in}$ are constant inputs to the populations. 

$c_2$ and $c_4$ are model parameters representing internal inhibitory coupling. 

$\tau_{ex}$ and $\tau_{in}$ are the poulation time constants.

$c_{EE}$ and $c_{EI}$ are self- and between population excitatory coupling parameters.

_pert_ represents external perturbation.

Function _sigmoid_ is a continuous step function, conmputationally implemented as tangens hyperbolicus.

```python

def plot_bifdiagram(results_min_f, results_max_f, 
                    par_set):
    
    N = len(results_min_f)

    fig, ax = subplots(nrows=2, figsize=(5, 4))

    for index in (0, 1):

        for xe, ye in zip(par_set, results_max_f[index]):
    
            if not isinstance(ye, ndarray):
                ax[index].scatter(xe, ye, c='r', s=6, marker='D')
            else:
                ax[index].scatter([xe] * len(ye), ye, s=3, c='k', marker='D')
    
        for xe, ye in zip(par_set, results_min_f[index]):
    
            if not isinstance(ye, ndarray):
                ax[index].scatter(xe, ye, c='r', s=6, marker='d')
            else:
                ax[index].scatter([xe] * len(ye), ye, s=3, c='gray', marker='d')
    
    
        ax[index].set_xticks(linspace(par_set[0], par_set[-1], 5));
        ax[index].set_xticklabels(around(linspace(par_set[0], par_set[-1], 5), 2), fontsize=16);
    
        ax[index].set_ylabel(f'EX_{index+1}', fontsize=14)
    
        y_min, y_max = ax[0].get_ylim()

        ax[index].set_ylim(y_min, y_max)
        ax[index].set_yticks(linspace(y_min, y_max, 3));
        ax[index].set_yticklabels(around(linspace(y_min, y_max, 3),2), fontsize=14);
        
    ax[1].set_xlabel('Parameter', fontsize=16)


    fig.tight_layout()
    
    return fig, ax
```

# Bifurcation Analysis

```python
import numpy as np
from scipy.integrate import solve_ivp
from scipy.signal import find_peaks
from scipy.interpolate import interp1d
import sk_dsp_comm.sigsys as ss
import matplotlib.pyplot as plt
from matplotlib.gridspec import GridSpec


def simulate_coupled_oscillators(pulse_fre, frac_E=0.2, frac_I=0.0,
                                 h_ex_0=-7., h_in_0=-4.0,
                                 pars=(1, 2.5, 10, 0, 5, 10),
                                 y_ini=(-1.9, -13.56, -1.9, -13.56),
                                 time_stop=150, sr=1000,
                                 pulse_wid=1., pulse_amp=1.2,
                                 seed=8888):
    """Simulate two bidirectionally coupled oscillators"""
    
    if y_ini is None:
        rng = np.random.default_rng(seed)
        y_ini = rng.uniform(size=4)
    
    time = np.linspace(start=0, stop=time_stop, num=time_stop*sr)
    pert = h_ex_0 + pulse_amp*ss.rect(np.mod(time, 1/pulse_fre)-(1/pulse_fre)/2+pulse_wid/2, pulse_wid)
    
    solution = solve_ivp(
        fun=two_oscillators, 
        t_span=(0, time_stop), 
        y0=y_ini,
        args=(h_ex_0, h_in_0, pars, frac_E, frac_I, sr, time_stop, pert),
        t_eval=time, 
        method='BDF', 
        max_step=0.1
    )
    
    return solution.t, solution.y.T, pert

def extract_all_peaks(t, y, transient=50, min_distance=500, height=None):
    """
    Extract ALL peak values from both oscillators after transient.
    Returns peaks for oscillator 1 (y[:, 0]) and oscillator 2 (y[:, 2]).
    """
    
    transient_idx = np.where(t > transient)[0]
    if len(transient_idx) == 0:
        return np.array([]), np.array([])
    
    transient_idx = transient_idx[0]
    
    # Find peaks for oscillator 1 (column 0)
    peaks_1, properties_1 = find_peaks(
        y[transient_idx:, 0], 
        distance=min_distance,
        height=height
    )
    peak_values_1 = y[transient_idx + peaks_1, 0]
    
    # Find peaks for oscillator 2 (column 2)
    peaks_2, properties_2 = find_peaks(
        y[transient_idx:, 2], 
        distance=min_distance,
        height=height
    )
    peak_values_2 = y[transient_idx + peaks_2, 2]
    
    return peak_values_1, peak_values_2

def extract_poincare_both_oscillators(t, y, pulse_fre, transient=50):
    """
    Extract Poincaré sections for both oscillators.
    Returns maximum values in each cycle for both.
    """
    
    cycle_starts = np.arange(0, t[-1], pulse_fre)
    cycle_starts = cycle_starts[cycle_starts > transient]
    
    poincare_max_1 = []
    poincare_max_2 = []
    
    for t_start in cycle_starts:
        t_end = t_start + pulse_fre
        cycle_idx = (t >= t_start) & (t < t_end)
        
        if np.any(cycle_idx):
            # Oscillator 1: y[0] = ex1
            poincare_max_1.append(np.max(y[cycle_idx, 0]))
            
            # Oscillator 2: y[2] = ex2
            poincare_max_2.append(np.max(y[cycle_idx, 2]))
    
    return cycle_starts, np.array(poincare_max_1), np.array(poincare_max_2)

def detect_propagation_failure(poincare_max_1, poincare_max_2, threshold=1.0):
    """
    Detect if oscillator 1 fires but oscillator 2 doesn't.
    """
    
    if len(poincare_max_1) != len(poincare_max_2):
        return False, 0
    
    # Count cycles where osc1 fires but osc2 doesn't
    osc1_fires = poincare_max_1 > threshold
    osc2_fires = poincare_max_2 > threshold
    
    propagation_failures = np.sum(osc1_fires & ~osc2_fires)
    total_osc1_fires = np.sum(osc1_fires)
    
    has_failure = propagation_failures > 0
    failure_rate = propagation_failures / total_osc1_fires if total_osc1_fires > 0 else 0
    
    return has_failure, failure_rate

def bifurcation_analysis_coupled(pulse_fres=None, 
                                frac_E_values=[0.2], 
                                frac_I_values=[0.0],
                                n_fres=20, 
                                **kwargs):
    """
    Bifurcation analysis for bidirectionally coupled oscillator system.
    Includes extraction of ALL peak values for classic bifurcation diagrams.
    """
    
    if pulse_fres is None:
        pulse_fres = np.linspace(0.4, 1.6, n_fres)
    
    # Store results for each coupling configuration
    all_results = {}
    
    for frac_E in frac_E_values:
        for frac_I in frac_I_values:
            coupling_key = (frac_E, frac_I)
            
            print(f"\n{'='*60}")
            print(f"Coupling: frac_E={frac_E:.3f}, frac_I={frac_I:.3f}")
            print('='*60)
            
            results = {
                'pulse_fres': pulse_fres,
                'frac_E': frac_E,
                'frac_I': frac_I,
                'response_std_1': [],
                'response_std_2': [],
                'response_ptp_1': [],
                'response_ptp_2': [],
                'poincare_max_1': [],
                'poincare_max_2': [],
                'all_peaks_1': [],  # NEW: All peak values for osc 1
                'all_peaks_2': [],  # NEW: All peak values for osc 2
                'mean_response_1': [],
                'mean_response_2': [],
                'n_peaks_1': [],
                'n_peaks_2': [],
                'propagation_failure': [],
                'failure_rate': []
            }
            
            for i, fre in enumerate(pulse_fres):
                if i % 10 == 0:
                    print(f"  Progress: {i}/{len(pulse_fres)} (freq={fre:.3f})")
                
                # Simulate
                t, y, pert = simulate_coupled_oscillators(
                    fre, frac_E=frac_E, frac_I=frac_I, **kwargs
                )
                
                # Extract ALL peaks for both oscillators
                peaks_1, peaks_2 = extract_all_peaks(t, y, transient=50, min_distance=500)
                
                # Extract Poincaré sections for both oscillators
                cycle_starts, poincare_max_1, poincare_max_2 = \
                    extract_poincare_both_oscillators(
                        t, y, kwargs.get('pulse_fre', 10)
                    )
                
                # Compute metrics for oscillator 1
                std_1 = np.std(poincare_max_1) if len(poincare_max_1) > 0 else 0.0
                ptp_1 = np.ptp(poincare_max_1) if len(poincare_max_1) > 0 else 0.0
                
                # Compute metrics for oscillator 2
                std_2 = np.std(poincare_max_2) if len(poincare_max_2) > 0 else 0.0
                ptp_2 = np.ptp(poincare_max_2) if len(poincare_max_2) > 0 else 0.0
                
                # Detect propagation failure
                has_failure, failure_rate = detect_propagation_failure(
                    poincare_max_1, poincare_max_2, threshold=1.0
                )
                
                # Store results
                results['response_std_1'].append(std_1)
                results['response_std_2'].append(std_2)
                results['response_ptp_1'].append(ptp_1)
                results['response_ptp_2'].append(ptp_2)
                results['poincare_max_1'].append(poincare_max_1)
                results['poincare_max_2'].append(poincare_max_2)
                results['all_peaks_1'].append(peaks_1)  # Store all peaks
                results['all_peaks_2'].append(peaks_2)  # Store all peaks
                results['mean_response_1'].append(
                    np.mean(poincare_max_1) if len(poincare_max_1) > 0 else 0
                )
                results['mean_response_2'].append(
                    np.mean(poincare_max_2) if len(poincare_max_2) > 0 else 0
                )
                results['propagation_failure'].append(has_failure)
                results['failure_rate'].append(failure_rate)
                results['n_peaks_1'].append(len(peaks_1))
                results['n_peaks_2'].append(len(peaks_2))
            
            # Convert to arrays (except all_peaks which are lists of arrays)
            for key in ['response_std_1', 'response_std_2', 'response_ptp_1', 
                        'response_ptp_2', 'mean_response_1', 'mean_response_2',
                        'n_peaks_1', 'n_peaks_2', 'propagation_failure', 'failure_rate']:
                results[key] = np.array(results[key])
            
            all_results[coupling_key] = results
    
    return all_results

def plot_classic_bifurcation_diagrams(results):
    """
    Classic bifurcation diagrams showing ALL peak values as points.
    Two separate plots: one for oscillator 1, one for oscillator 2.
    """
    
    pulse_fres = results['pulse_fres']
    frac_E = results['frac_E']
    frac_I = results['frac_I']
    
    fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(14, 6), sharex=True)
    
    # 1. Oscillator 1: Plot all peak values
    for i, fre in enumerate(pulse_fres):
        peaks = results['all_peaks_1'][i]
        if len(peaks) > 0:
            ax1.plot([fre]*len(peaks), peaks, 'b.', markersize=3, alpha=0.6)
    
    ax1.set_ylabel('Oscillator 1 Peaks', fontsize=13)
    ax1.set_title(f'Oscillator 1 (Stimulated)', 
                  fontsize=13, weight='bold')
    ax1.grid(True, alpha=0.3)
    
    # Mark propagation failure regions
    if np.any(results['propagation_failure']):
        failure_fres = pulse_fres[results['propagation_failure']]
        y_lim = ax1.get_ylim()
        for fre in failure_fres:
            ax1.axvline(fre, color='red', alpha=0.1, linewidth=3)
        # ax1.text(0.98, 0.97, 'Red shading:\nPropagation failure', 
        #         transform=ax1.transAxes, ha='right', va='top',
        #         fontsize=9, bbox=dict(boxstyle='round', facecolor='lightyellow', alpha=0.7))
    
    # 2. Oscillator 2: Plot all peak values
    for i, fre in enumerate(pulse_fres):
        peaks = results['all_peaks_2'][i]
        if len(peaks) > 0:
            ax2.plot([fre]*len(peaks), peaks, 'r.', markersize=3, alpha=0.6)
    
    ax2.set_xlabel('Pulse Frequency', fontsize=13)
    ax2.set_ylabel('Oscillator 2 Peaks', fontsize=13)
    ax2.set_title(f'Oscillator 2', 
                  fontsize=13, weight='bold')
    ax2.grid(True, alpha=0.3)
    
    # Mark propagation failure regions
    if np.any(results['propagation_failure']):
        failure_fres = pulse_fres[results['propagation_failure']]
        for fre in failure_fres:
            ax2.axvline(fre, color='red', alpha=0.1, linewidth=2)
    
    plt.suptitle('Bifurcation Diagrams', 
                 fontsize=16, weight='bold', y=0.995)
    plt.tight_layout()
    
    return fig

def plot_combined_bifurcation_overview(results):
    """
    Comprehensive 5-panel view:
    1. Classic bifurcation diagram for oscillator 1 (all peaks)
    2. Classic bifurcation diagram for oscillator 2 (all peaks)
    3. Response statistics (std)
    4. Difference analysis
    5. Propagation failure rate
    """
    
    pulse_fres = results['pulse_fres']
    frac_E     = results['frac_E']
    frac_I     = results['frac_I']

    fig, ax = plt.subplots(figsize=(14, 4))

    
    # fig = plt.figure(figsize=(16, 5))
    # gs = GridSpec(1, 1, figure=fig, hspace=0.35, wspace=0.3)
    
    # # 1. Oscillator 1: Classic bifurcation (all peaks)
    # ax1 = fig.add_subplot(gs[0, :])
    # for i, amp in enumerate(pulse_amps):
    #     peaks = results['all_peaks_1'][i]
    #     if len(peaks) > 0:
    #         ax1.plot([amp]*len(peaks), peaks, 'b.', markersize=4, alpha=0.6)
    
    # ax1.set_ylabel('Osc 1 Peak Values', fontsize=12)
    # ax1.set_title(f'Oscillator 1: All Peaks (Stimulated)', fontsize=13, weight='bold')
    # ax1.grid(True, alpha=0.3)
    
    # # Mark propagation failures
    # if np.any(results['propagation_failure']):
    #     for amp in pulse_amps[results['propagation_failure']]:
    #         ax1.axvline(amp, color='red', alpha=0.1, linewidth=2)
    
    # # 2. Oscillator 2: Classic bifurcation (all peaks)
    # ax2 = fig.add_subplot(gs[1, :], sharex=ax1)
    # for i, amp in enumerate(pulse_amps):
    #     peaks = results['all_peaks_2'][i]
    #     if len(peaks) > 0:
    #         ax2.plot([amp]*len(peaks), peaks, 'r.', markersize=4, alpha=0.6)
    
    # ax2.set_ylabel('Osc 2 Peak Values', fontsize=12)
    # ax2.set_title(f'Oscillator 2: All Peaks (Coupled)', fontsize=13, weight='bold')
    # ax2.grid(True, alpha=0.3)
    
    # # Mark propagation failures
    # if np.any(results['propagation_failure']):
    #     for amp in pulse_amps[results['propagation_failure']]:
    #         ax2.axvline(amp, color='red', alpha=0.1, linewidth=2)
    
    # # 3. Response statistics (std)
    # ax3 = fig.add_subplot(gs[0, 0], sharex=None)
    # ax3.plot(pulse_amps, results['response_std_1'], 'b-', linewidth=2, label='Osc 1', alpha=0.8)
    # ax3.plot(pulse_amps, results['response_std_2'], 'r-', linewidth=2, label='Osc 2', alpha=0.8)
    # ax3.set_ylabel('Response Amplitude (Std)', fontsize=11)
    # ax3.set_title('Response Statistics', fontsize=12, weight='bold')
    # ax3.legend(fontsize=10)
    # ax3.grid(True, alpha=0.3)
    
    # # 4. Number of peaks
    # ax4 = fig.add_subplot(gs, sharex=None)
    # ax4.plot(pulse_fres, results['n_peaks_1'], 'b-', linewidth=2, label='Osc 1', alpha=0.8)
    # ax4.plot(pulse_fres, results['n_peaks_2'], 'r-', linewidth=2, label='Osc 2', alpha=0.8)
    # ax4.set_ylabel('Number of Peaks', fontsize=11)
    # ax4.set_title('Peak Count', fontsize=12, weight='bold')
    # ax4.legend(fontsize=10)
    # ax4.grid(True, alpha=0.3)
    
    # # 5. Difference
    # ax5 = fig.add_subplot(gs[1, 0], sharex=None)
    # difference = results['response_std_1'] - results['response_std_2']
    # ax5.plot(pulse_amps, difference, 'purple', linewidth=2)
    # ax5.fill_between(pulse_amps, 0, difference, 
    #                  where=(difference > 0), alpha=0.3, color='purple', label='Osc1 > Osc2')
    # ax5.fill_between(pulse_amps, 0, difference, 
    #                  where=(difference < 0), alpha=0.3, color='orange', label='Osc2 > Osc1')
    # ax5.axhline(0, color='k', linestyle='--', alpha=0.5)
    # ax5.set_xlabel('Pulse Amplitude', fontsize=12)
    # ax5.set_ylabel('Response Difference', fontsize=11)
    # ax5.set_title('Osc1 - Osc2', fontsize=12, weight='bold')
    # ax5.legend(fontsize=10)
    # ax5.grid(True, alpha=0.3)
    
    # 6. Propagation failure rate
    # ax = fig.add_subplot(gs[0, 1], sharex=None)
    ax.plot(pulse_fres, results['failure_rate'] * 100, 
             'darkred', linewidth=2, marker='o', markersize=4)
    ax.fill_between(pulse_fres, 0, results['failure_rate'] * 100, alpha=0.3, color='red')
    ax.set_xlabel('Pulse Frequency', fontsize=12)
    ax.set_ylabel('Failure Rate (%)', fontsize=11)
    ax.set_title('Propagation Failure', fontsize=12, weight='bold')
    ax.set_ylim([-5, 105])
    ax.grid(True, alpha=0.3)
    
    # plt.suptitle(f'Comprehensive Bifurcation Analysis (frac_E={frac_E:.2f}, frac_I={frac_I:.2f})', 
    #              fontsize=16, weight='bold', y=0.997)
    
    return fig

def plot_timeseries_comparison(pulse_fres=[0.6, 0.8, 1.0], 
                               frac_E=0.2, frac_I=0.0, **kwargs):
    """
    Compare time series of both oscillators at different pulse frequencies.
    """
    
    n = len(pulse_fres)
    fig, axes = plt.subplots(1, n, figsize=(16, 5), sharex=True)
    if n == 1:
        axes = [axes]
    
    for i, fre in enumerate(pulse_fres):
        t, y, pert = simulate_coupled_oscillators(
            fre, frac_E=frac_E, frac_I=frac_I, **kwargs
        )
        
        # Plot last portion to see steady state
        idx = t > max(100, t[-1] - 51)
        
        ax = axes[i]
        
        # Oscillator 1
        ax.plot(t[idx], y[idx, 0], 'b-', linewidth=2, 
               label='Osc1 (Ex)', alpha=0.8)
        
        # Oscillator 2
        ax.plot(t[idx], y[idx, 2], 'r-', linewidth=2, 
               label='Osc2 (Ex)', alpha=0.8)
        
        # # Perturbation
        # ax.plot(t[idx], pert[idx], 'k--', linewidth=1, 
        #        alpha=0.3, label='Perturbation')
        
        ax.set_ylabel(f'Freq={fre:.2f}', fontsize=12, weight='bold')
        ax.legend(loc='upper right', fontsize=10)
        ax.grid(True, alpha=0.3)
        
        # Detect if there's propagation failure
        _, poincare_1, poincare_2 = extract_poincare_both_oscillators(
            t, y, kwargs.get('pulse_fre', 10)
        )
        has_failure, failure_rate = detect_propagation_failure(poincare_1, poincare_2)
        
        if has_failure:
            ax.set_facecolor('#fff0f0')
            ax.text(0.02, 0.95, f'PROPAGATION FAILURE ({failure_rate*100:.0f}%)', 
                   transform=ax.transAxes, fontsize=11, weight='bold', 
                   color='red', va='top')
        elif np.std(poincare_2) > 0.5:  # Osc2 is responding
            ax.set_facecolor('#f0fff0')
            ax.text(0.02, 0.95, '✓ Successful propagation', 
                   transform=ax.transAxes, fontsize=10, 
                   style='italic', color='green', va='top')
        # else:  # Both quiescent
        #     ax.set_facecolor('#f0f0f0')
        #     ax.text(0.02, 0.95, 'Both quiescent', 
        #            transform=ax.transAxes, fontsize=10, 
        #            style='italic', color='gray', va='top')
    
        ax.set_xlabel('Time', fontsize=13)
        ax.set_ylim(-1.8, 2.6)
        
    plt.suptitle(f'Time Series (frac_E={frac_E:.2f}, frac_I={frac_I:.2f})', 
                 fontsize=14, weight='bold')
    plt.tight_layout()
    
    return fig

def plot_state_space_comparison(pulse_fres=[0.6, 0.9, 1.2], 
                                frac_E=0.2, frac_I=0.0, **kwargs):
    """
    State space plots showing both oscillators.
    """
    
    n = len(pulse_fres)
    fig, axes = plt.subplots(1, n, figsize=(6*n, 5))
    if n == 1:
        axes = [axes]
    
    for i, fre in enumerate(pulse_fres):
        t, y, pert = simulate_coupled_oscillators(
            fre, frac_E=frac_E, frac_I=frac_I, **kwargs
        )
        
        ax = axes[i]
        
        # Plot after transient
        transient_idx = int(50 * kwargs.get('sr', 1000))
        
        # Oscillator 1 trajectory (y[0] vs y[1])
        ax.plot(y[transient_idx:, 0], y[transient_idx:, 1], 
               'b-', linewidth=1.5, alpha=0.6, label='Osc1')
        
        # Oscillator 2 trajectory (y[2] vs y[3])
        ax.plot(y[transient_idx:, 2], y[transient_idx:, 3], 
               'r-', linewidth=1.5, alpha=0.6, label='Osc2')
        
        # Mark initial points
        ax.plot(y[transient_idx, 0], y[transient_idx, 1], 
               'bo', markersize=8)
        ax.plot(y[transient_idx, 2], y[transient_idx, 3], 
               'ro', markersize=8)
        
        ax.set_xlabel('Excitatory', fontsize=12)
        ax.set_ylabel('Inhibitory', fontsize=12)
        ax.set_title(f'Freq = {fre:.3f}', fontsize=12, weight='bold')
        ax.legend(fontsize=9, loc='best')
        ax.grid(True, alpha=0.3)
        ax.set_ylim(-14, 2)
        
        # Detect propagation failure
        _, poincare_1, poincare_2 = extract_poincare_both_oscillators(
            t, y, kwargs.get('pulse_fre', 10)
        )
        has_failure, _ = detect_propagation_failure(poincare_1, poincare_2)
        
        if has_failure:
            ax.set_facecolor('#fff0f0')
        elif np.std(poincare_2) > 0.5:
            ax.set_facecolor('#f0fff0')
        else:
            ax.set_facecolor('#f0f0f0')
    
    plt.suptitle(f'State Space (frac_E={frac_E:.2f}, frac_I={frac_I:.2f})', 
                 fontsize=14, weight='bold')
    plt.tight_layout()
    
    return fig

# Main execution
if __name__ == "__main__":
    print("="*70)
    print("BIDIRECTIONALLY COUPLED OSCILLATORS: BIFURCATION ANALYSIS")
    print("="*70)
    
    # Parameters
    pulse_fres = np.linspace(0.5, 1.5, 200)
    
    # Single coupling configuration
    print("\n--- Single coupling configuration ---")
    frac_E = 0.2
    frac_I = 0.0
    
    all_results = bifurcation_analysis_coupled(
        # pulse_amps=pulse_amps,
        pulse_fres=pulse_fres,
        frac_E_values=[frac_E],
        frac_I_values=[frac_I],
        time_stop=150,
        sr=1000,
        pulse_wid=1.0,
        pulse_amp=1.2,
        seed=88888
    )
    
    results = all_results[(frac_E, frac_I)]
    
    # NEW: Classic bifurcation diagrams with all peaks
    fig1 = plot_classic_bifurcation_diagrams(results)
    plt.savefig(f'classic_bifurcation_fE{frac_E:.2f}_fI{frac_I:.2f}.png', 
                dpi=150, bbox_inches='tight')
    
    # NEW: Comprehensive overview
    fig2 = plot_combined_bifurcation_overview(results)
    plt.savefig(f'comprehensive_bifurcation_fE{frac_E:.2f}_fI{frac_I:.2f}.png', 
                dpi=150, bbox_inches='tight')
    
    # Plot time series examples
    fig3 = plot_timeseries_comparison(
        pulse_fres=[0.8, 1.0, 1.2],
        frac_E=frac_E, frac_I=frac_I,
        time_stop=150, sr=1000, pulse_wid=1., pulse_amp=1.2, seed=8888
    )
    plt.savefig(f'timeseries_fE{frac_E:.2f}_fI{frac_I:.2f}.png', 
                dpi=150, bbox_inches='tight')
    
    # Plot state space
    fig4 = plot_state_space_comparison(
        pulse_fres=[0.8, 1.0, 1.2],
        frac_E=frac_E, frac_I=frac_I,
        time_stop=150, sr=1000, pulse_wid=1., pulse_amp=1.2, seed=8888
    )
    plt.savefig(f'state_space_fE{frac_E:.2f}_fI{frac_I:.2f}.png', 
                dpi=150, bbox_inches='tight')
    
    plt.show()
    
    # Summary
    print("\n" + "="*70)
    print("SUMMARY")
    print("="*70)
    print(f"Coupling configuration: frac_E={frac_E:.3f}, frac_I={frac_I:.3f}")
    
    if np.any(results['propagation_failure']):
        failure_fres = pulse_fres[results['propagation_failure']]
        print(f"\n*** PROPAGATION FAILURE DETECTED ***")
        print(f"Failure in frequency range: [{failure_fres.min():.3f}, {failure_fres.max():.3f}]")
        print(f"Maximum failure rate: {results['failure_rate'].max()*100:.1f}%")
    else:
        print("\n*** NO PROPAGATION FAILURE ***")
        print("Excitation successfully propagates")
    
    # Peak statistics
    total_peaks_1 = sum(len(p) for p in results['all_peaks_1'])
    total_peaks_2 = sum(len(p) for p in results['all_peaks_2'])
    print(f"\nTotal peaks found:")
    print(f"  Oscillator 1: {total_peaks_1}")
    print(f"  Oscillator 2: {total_peaks_2}")
```

```python
# import numpy as np
# from scipy.integrate import solve_ivp
# from scipy.signal import find_peaks
# from scipy.interpolate import interp1d
# import sk_dsp_comm.sigsys as ss
# import matplotlib.pyplot as plt
# from matplotlib.gridspec import GridSpec

# def sigmoid(u):
#     return np.tanh(u)

# def two_oscillators(t, y, h_ex, h_in, pars, frac_E, frac_I, sr, time_stop, pert):
#     """
#     Two bidirectionally coupled oscillators.
#     Coupling enters through sigmoid arguments: sigmoid(y[0] + frac_E*y[2])
#     """
#     tau_ex, tau_in, c2, c4, c_EE, c_EI = pars
#     time_index = int(t*sr)

#     if time_index >= time_stop*sr:
#         dydt = np.zeros(4)
#         return dydt
#     else:
#         dydt = (
#             (pert[time_index] - y[0] - c2*sigmoid(y[1]) + c_EE*sigmoid(y[0]+frac_E*y[2]))*tau_ex,
#             (h_in             - y[1] - c4*sigmoid(y[1]) + c_EI*sigmoid(y[0]+frac_I*y[2]))*tau_in,
#             (h_ex             - y[2] - c2*sigmoid(y[3]) + c_EE*sigmoid(y[2]+frac_E*y[0]))*tau_ex,
#             (h_in             - y[3] - c4*sigmoid(y[3]) + c_EI*sigmoid(y[2]+frac_I*y[0]))*tau_in
#         )
#     return dydt

# def simulate_coupled_oscillators(pulse_amp, pulse_per=10, frac_E=0.2, frac_I=0.0,
#                                  h_ex_0=-6.72, h_in_0=-4.0,
#                                  pars=(1, 2.5, 10, 0, 5, 10),
#                                  y_ini=None,
#                                  time_stop=100, sr=1000,
#                                  pulse_wid=4.,
#                                  seed=8888):
#     """Simulate two bidirectionally coupled oscillators"""
    
#     if y_ini is None:
#         rng = np.random.default_rng(seed)
#         y_ini = rng.uniform(size=4)
    
#     time = np.linspace(start=0, stop=time_stop, num=time_stop*sr)
#     pert = h_ex_0 + pulse_amp*ss.rect(
#         np.mod(time, pulse_per)-(pulse_per)/2+pulse_wid/2, pulse_wid)
    
#     solution = solve_ivp(
#         fun=two_oscillators, 
#         t_span=(0, time_stop), 
#         y0=y_ini,
#         args=(h_ex_0, h_in_0, pars, frac_E, frac_I, sr, time_stop, pert),
#         t_eval=time, 
#         method='BDF', 
#         max_step=0.1
#     )
    
#     return solution.t, solution.y.T, pert

def extract_all_peaks(t, y, transient=50, min_distance=50, height=None, threshold=0.0):
    """
    Extract ALL peak values from both oscillators after transient.
    Only returns peaks with values > threshold (default 0).
    This filters out subthreshold oscillations.
    
    Returns peaks for oscillator 1 (y[:, 0]) and oscillator 2 (y[:, 2]).
    """
    
    transient_idx = np.where(t > transient)[0]
    if len(transient_idx) == 0:
        return np.array([]), np.array([])
    
    transient_idx = transient_idx[0]
    
    # Find peaks for oscillator 1 (column 0)
    peaks_1, properties_1 = find_peaks(
        y[transient_idx:, 0], 
        distance=min_distance,
        height=height
    )
    peak_values_1 = y[transient_idx + peaks_1, 0]
    
    # Filter for peaks > threshold
    valid_peaks_1 = peak_values_1 > threshold
    peak_values_1 = peak_values_1[valid_peaks_1]
    
    # Find peaks for oscillator 2 (column 2)
    peaks_2, properties_2 = find_peaks(
        y[transient_idx:, 2], 
        distance=min_distance,
        height=height
    )
    peak_values_2 = y[transient_idx + peaks_2, 2]
    
    # Filter for peaks > threshold
    valid_peaks_2 = peak_values_2 > threshold
    peak_values_2 = peak_values_2[valid_peaks_2]
    
    return peak_values_1, peak_values_2

def extract_poincare_both_oscillators(t, y, pulse_per, transient=50):
    """
    Extract Poincaré sections for both oscillators.
    Returns maximum values in each cycle for both.
    """
    
    cycle_starts = np.arange(0, t[-1], pulse_per)
    cycle_starts = cycle_starts[cycle_starts > transient]
    
    poincare_max_1 = []
    poincare_max_2 = []
    
    for t_start in cycle_starts:
        t_end = t_start + pulse_per
        cycle_idx = (t >= t_start) & (t < t_end)
        
        if np.any(cycle_idx):
            # Oscillator 1: y[0] = ex1
            poincare_max_1.append(np.max(y[cycle_idx, 0]))
            
            # Oscillator 2: y[2] = ex2
            poincare_max_2.append(np.max(y[cycle_idx, 2]))
    
    return cycle_starts, np.array(poincare_max_1), np.array(poincare_max_2)

def detect_propagation_failure(poincare_max_1, poincare_max_2, threshold=0.0):
    """
    Detect if oscillator 1 fires but oscillator 2 doesn't.
    """
    
    if len(poincare_max_1) != len(poincare_max_2):
        return False, 0
    
    # Count cycles where osc1 fires but osc2 doesn't
    osc1_fires = poincare_max_1 > threshold
    osc2_fires = poincare_max_2 > threshold
    
    propagation_failures = np.sum(osc1_fires & ~osc2_fires)
    total_osc1_fires = np.sum(osc1_fires)
    
    has_failure = propagation_failures > 0
    failure_rate = propagation_failures / total_osc1_fires if total_osc1_fires > 0 else 0
    
    return has_failure, failure_rate

def bifurcation_analysis_frequency_amplitude(pulse_amps=None,
                                            pulse_per_values=None,
                                            frac_E=0.2, 
                                            frac_I=0.0,
                                            n_amps=20,
                                            n_freqs=20,
                                            peak_threshold=0.0,  # NEW parameter
                                            **kwargs):
    """
    Frequency-amplitude bifurcation analysis (Arnold tongue diagram).
    Scans over pulse_amp (amplitude) and pulse_per (period/frequency).
    Fixed coupling: frac_E and frac_I.
    """
    
    if pulse_amps is None:
        pulse_amps = np.linspace(0.0, 1.0, n_amps)
    
    if pulse_per_values is None:
        pulse_per_values = np.linspace(1, 10, n_freqs)
    
    print(f"\n{'='*70}")
    print(f"FREQUENCY-AMPLITUDE BIFURCATION ANALYSIS")
    print(f"Coupling: frac_E={frac_E:.3f}, frac_I={frac_I:.3f}")
    print(f"Peak threshold: {peak_threshold:.2f}")
    print(f"Scanning {len(pulse_amps)} amplitudes × {len(pulse_per_values)} periods")
    print('='*70)
    
    # Store results for all parameter combinations
    results = {
        'pulse_amps': pulse_amps,
        'pulse_per_values': pulse_per_values,
        'frac_E': frac_E,
        'frac_I': frac_I,
        'peak_threshold': peak_threshold,
        'response_std_1': np.zeros((len(pulse_per_values), len(pulse_amps))),
        'response_std_2': np.zeros((len(pulse_per_values), len(pulse_amps))),
        'all_peaks_1': [[None for _ in pulse_amps] for _ in pulse_per_values],
        'all_peaks_2': [[None for _ in pulse_amps] for _ in pulse_per_values],
        'n_peaks_1': np.zeros((len(pulse_per_values), len(pulse_amps))),
        'n_peaks_2': np.zeros((len(pulse_per_values), len(pulse_amps))),
        'propagation_failure': np.zeros((len(pulse_per_values), len(pulse_amps)), dtype=bool),
        'failure_rate': np.zeros((len(pulse_per_values), len(pulse_amps)))
    }
    
    total_sims = len(pulse_per_values) * len(pulse_amps)
    sim_count = 0
    
    for i, pulse_per in enumerate(pulse_per_values):
        print(f"\nPeriod {i+1}/{len(pulse_per_values)}: pulse_per = {pulse_per:.2f}")
        
        for j, amp in enumerate(pulse_amps):
            sim_count += 1
            if sim_count % 50 == 0:
                print(f"  Progress: {sim_count}/{total_sims} ({100*sim_count/total_sims:.1f}%)")
            
            # Simulate with this period and amplitude
            t, y, pert = simulate_coupled_oscillators(
                amp, pulse_per=pulse_per, frac_E=frac_E, frac_I=frac_I, **kwargs
            )
            
            # Extract ALL peaks for both oscillators (with threshold filter)
            peaks_1, peaks_2 = extract_all_peaks(
                t, y, transient=50, min_distance=500, threshold=peak_threshold
            )
            
            # Extract Poincaré sections
            cycle_starts, poincare_max_1, poincare_max_2 = \
                extract_poincare_both_oscillators(t, y, pulse_per)
            
            # Compute metrics
            std_1 = np.std(poincare_max_1) if len(poincare_max_1) > 0 else 0.0
            std_2 = np.std(poincare_max_2) if len(poincare_max_2) > 0 else 0.0
            
            # Detect propagation failure (using peak threshold)
            has_failure, failure_rate = detect_propagation_failure(
                poincare_max_1, poincare_max_2, threshold=peak_threshold
            )
            
            # Store results
            results['response_std_1'][i, j] = std_1
            results['response_std_2'][i, j] = std_2
            results['all_peaks_1'][i][j] = peaks_1
            results['all_peaks_2'][i][j] = peaks_2
            results['n_peaks_1'][i, j] = len(peaks_1)
            results['n_peaks_2'][i, j] = len(peaks_2)
            results['propagation_failure'][i, j] = has_failure
            results['failure_rate'][i, j] = failure_rate
    
    print(f"\n{'='*70}")
    print("Analysis complete!")
    print('='*70)
    
    return results

def plot_frequency_amplitude_diagram(results):
    """
    2D frequency-amplitude bifurcation diagram (Arnold tongue style).
    Shows response amplitude as color/intensity.
    """
    
    pulse_amps = results['pulse_amps']
    pulse_per_values = results['pulse_per_values']
    frac_E = results['frac_E']
    frac_I = results['frac_I']
    
    # Convert period to frequency for more intuitive plotting
    frequencies = 1.0 / pulse_per_values
    
    fig, axes = plt.subplots(2, 2, figsize=(16, 12))
    
    # 1. Oscillator 1 response amplitude
    im1 = axes[0, 0].imshow(results['response_std_1'], 
                            aspect='auto', origin='lower',
                            extent=[pulse_amps[0], pulse_amps[-1], 
                                   frequencies[0], frequencies[-1]],
                            cmap='viridis', interpolation='bilinear')
    axes[0, 0].set_xlabel('Pulse Amplitude', fontsize=12)
    axes[0, 0].set_ylabel('Pulse Frequency (1/period)', fontsize=12)
    axes[0, 0].set_title('Oscillator 1: Response Amplitude', fontsize=13, weight='bold')
    plt.colorbar(im1, ax=axes[0, 0], label='Std Dev')
    
    # 2. Oscillator 2 response amplitude
    im2 = axes[0, 1].imshow(results['response_std_2'], 
                            aspect='auto', origin='lower',
                            extent=[pulse_amps[0], pulse_amps[-1], 
                                   frequencies[0], frequencies[-1]],
                            cmap='viridis', interpolation='bilinear')
    axes[0, 1].set_xlabel('Pulse Amplitude', fontsize=12)
    axes[0, 1].set_ylabel('Pulse Frequency (1/period)', fontsize=12)
    axes[0, 1].set_title('Oscillator 2: Response Amplitude', fontsize=13, weight='bold')
    plt.colorbar(im2, ax=axes[0, 1], label='Std Dev')
    
    # 3. Number of peaks (entrainment/complexity)
    im3 = axes[1, 0].imshow(results['n_peaks_1'], 
                            aspect='auto', origin='lower',
                            extent=[pulse_amps[0], pulse_amps[-1], 
                                   frequencies[0], frequencies[-1]],
                            cmap='hot', interpolation='nearest')
    axes[1, 0].set_xlabel('Pulse Amplitude', fontsize=12)
    axes[1, 0].set_ylabel('Pulse Frequency (1/period)', fontsize=12)
    axes[1, 0].set_title('Oscillator 1: Number of Peaks', fontsize=13, weight='bold')
    plt.colorbar(im3, ax=axes[1, 0], label='Peak Count')

    # 4. Number of peaks (entrainment/complexity)
    im4 = axes[1, 1].imshow(results['n_peaks_2'], 
                            aspect='auto', origin='lower',
                            extent=[pulse_amps[0], pulse_amps[-1], 
                                   frequencies[0], frequencies[-1]],
                            cmap='hot', interpolation='nearest')
    axes[1, 1].set_xlabel('Pulse Amplitude', fontsize=12)
    axes[1, 1].set_ylabel('Pulse Frequency (1/period)', fontsize=12)
    axes[1, 1].set_title('Oscillator 1: Number of Peaks', fontsize=13, weight='bold')
    plt.colorbar(im4, ax=axes[1, 1], label='Peak Count')


    # # 4. Propagation failure rate
    # im4 = axes[1, 1].imshow(results['failure_rate'] * 100, 
    #                         aspect='auto', origin='lower',
    #                         extent=[pulse_amps[0], pulse_amps[-1], 
    #                                frequencies[0], frequencies[-1]],
    #                         cmap='Reds', vmin=0, vmax=100, interpolation='bilinear')
    # axes[1, 1].set_xlabel('Pulse Amplitude', fontsize=12)
    # axes[1, 1].set_ylabel('Pulse Frequency (1/period)', fontsize=12)
    # axes[1, 1].set_title('Propagation Failure Rate', fontsize=13, weight='bold')
    # plt.colorbar(im4, ax=axes[1, 1], label='Failure Rate (%)')

    
    
    plt.suptitle(f'Frequency-Amplitude Response Diagram (frac_E={frac_E:.2f}, frac_I={frac_I:.2f})', 
                 fontsize=16, weight='bold')
    plt.tight_layout()
    
    return fig

def plot_classic_bifurcation_slices(results, slice_periods=[1, 3, 5, 10]):
    """
    Classic bifurcation diagrams (all peaks) for selected pulse periods.
    Shows how the bifurcation structure changes with driving frequency.
    """
    
    pulse_amps = results['pulse_amps']
    pulse_per_values = results['pulse_per_values']
    
    n = len(slice_periods)
    fig, axes = plt.subplots(2, n, figsize=(5*n, 10), sharex=True)
    
    for col, pulse_per in enumerate(slice_periods):
        # Find closest period in our data
        per_idx = np.argmin(np.abs(pulse_per_values - pulse_per))
        actual_per = pulse_per_values[per_idx]
        
        # Oscillator 1
        ax1 = axes[0, col]
        for j, amp in enumerate(pulse_amps):
            peaks = results['all_peaks_1'][per_idx][j]
            if peaks is not None and len(peaks) > 0:
                ax1.plot([amp]*len(peaks), peaks, 'b.', markersize=3, alpha=0.6)
        
        ax1.set_ylabel('Osc 1 Peak Values', fontsize=11)
        ax1.set_title(f'Period = {actual_per:.2f}\n(Freq = {1/actual_per:.3f})', 
                     fontsize=11, weight='bold')
        ax1.grid(True, alpha=0.3)
        
        # Oscillator 2
        ax2 = axes[1, col]
        for j, amp in enumerate(pulse_amps):
            peaks = results['all_peaks_2'][per_idx][j]
            if peaks is not None and len(peaks) > 0:
                ax2.plot([amp]*len(peaks), peaks, 'r.', markersize=3, alpha=0.6)
        
        ax2.set_xlabel('Pulse Amplitude', fontsize=11)
        ax2.set_ylabel('Osc 2 Peak Values', fontsize=11)
        ax2.grid(True, alpha=0.3)
    
    axes[0, 0].text(-0.15, 0.5, 'Oscillator 1', transform=axes[0, 0].transAxes,
                    rotation=90, va='center', fontsize=13, weight='bold')
    axes[1, 0].text(-0.15, 0.5, 'Oscillator 2', transform=axes[1, 0].transAxes,
                    rotation=90, va='center', fontsize=13, weight='bold')
    
    plt.suptitle('Bifurcation Diagrams at Different Driving Frequencies', 
                 fontsize=14, weight='bold')
    plt.tight_layout()
    
    return fig

def plot_amplitude_slices(results, slice_amplitudes=[0.0, 0.3, 0.6, 0.9]):
    """
    Response vs frequency for fixed amplitudes.
    Classic resonance curves.
    """
    
    pulse_amps = results['pulse_amps']
    pulse_per_values = results['pulse_per_values']
    frequencies = 1.0 / pulse_per_values
    
    fig, axes = plt.subplots(2, 2, figsize=(14, 10))
    
    colors = plt.cm.viridis(np.linspace(0, 1, len(slice_amplitudes)))
    
    # 1. Oscillator 1 response vs frequency
    ax1 = axes[0, 0]
    for amp, color in zip(slice_amplitudes, colors):
        amp_idx = np.argmin(np.abs(pulse_amps - amp))
        actual_amp = pulse_amps[amp_idx]
        ax1.plot(frequencies, results['response_std_1'][:, amp_idx], 
                'o-', color=color, linewidth=2, markersize=5, 
                label=f'A={actual_amp:.2f}', alpha=0.8)
    
    ax1.set_xlabel('Pulse Frequency', fontsize=12)
    ax1.set_ylabel('Osc 1 Response (Std)', fontsize=12)
    ax1.set_title('Oscillator 1: Resonance Curves', fontsize=12, weight='bold')
    ax1.legend(fontsize=9)
    ax1.grid(True, alpha=0.3)
    
    # 2. Oscillator 2 response vs frequency
    ax2 = axes[0, 1]
    for amp, color in zip(slice_amplitudes, colors):
        amp_idx = np.argmin(np.abs(pulse_amps - amp))
        actual_amp = pulse_amps[amp_idx]
        ax2.plot(frequencies, results['response_std_2'][:, amp_idx], 
                'o-', color=color, linewidth=2, markersize=5, 
                label=f'A={actual_amp:.2f}', alpha=0.8)
    
    ax2.set_xlabel('Pulse Frequency', fontsize=12)
    ax2.set_ylabel('Osc 2 Response (Std)', fontsize=12)
    ax2.set_title('Oscillator 2: Resonance Curves', fontsize=12, weight='bold')
    ax2.legend(fontsize=9)
    ax2.grid(True, alpha=0.3)
    
    # 3. Number of peaks vs frequency
    ax3 = axes[1, 0]
    for amp, color in zip(slice_amplitudes, colors):
        amp_idx = np.argmin(np.abs(pulse_amps - amp))
        actual_amp = pulse_amps[amp_idx]
        ax3.plot(frequencies, results['n_peaks_1'][:, amp_idx], 
                'o-', color=color, linewidth=2, markersize=5, 
                label=f'A={actual_amp:.2f}', alpha=0.8)
    
    ax3.set_xlabel('Pulse Frequency', fontsize=12)
    ax3.set_ylabel('Osc 1 Peak Count', fontsize=12)
    ax3.set_title('Entrainment/Complexity vs Frequency', fontsize=12, weight='bold')
    ax3.legend(fontsize=9)
    ax3.grid(True, alpha=0.3)
    
    # 4. Propagation failure vs frequency
    ax4 = axes[1, 1]
    for amp, color in zip(slice_amplitudes, colors):
        amp_idx = np.argmin(np.abs(pulse_amps - amp))
        actual_amp = pulse_amps[amp_idx]
        ax4.plot(frequencies, results['failure_rate'][:, amp_idx] * 100, 
                'o-', color=color, linewidth=2, markersize=5, 
                label=f'A={actual_amp:.2f}', alpha=0.8)
    
    ax4.set_xlabel('Pulse Frequency', fontsize=12)
    ax4.set_ylabel('Failure Rate (%)', fontsize=12)
    ax4.set_title('Propagation Failure vs Frequency', fontsize=12, weight='bold')
    ax4.legend(fontsize=9)
    ax4.grid(True, alpha=0.3)
    ax4.set_ylim([-5, 105])
    
    plt.suptitle('Response Characteristics vs Driving Frequency', 
                 fontsize=14, weight='bold')
    plt.tight_layout()
    
    return fig

# Main execution
if __name__ == "__main__":
    print("="*70)
    print("FREQUENCY-AMPLITUDE BIFURCATION ANALYSIS")
    print("Classic Arnold Tongue / Resonance Diagram")
    print("="*70)
    
    # Parameters for frequency-amplitude scan
    pulse_amps       = np.linspace(0., 0.6, 25)  # Amplitude axis
    pulse_per_values = np.linspace(1., 10, 25)  # Period axis (frequency = 1/period)
    
    frac_E = 0.2
    frac_I = 0.0
    peak_threshold = 0.0  # Only count peaks > 0
    
    # Run frequency-amplitude analysis
    results = bifurcation_analysis_frequency_amplitude(
        pulse_amps=pulse_amps,
        pulse_per_values=pulse_per_values,
        frac_E=frac_E,
        frac_I=frac_I,
        peak_threshold=peak_threshold,
        time_stop=150,
        sr=1000,
        pulse_wid=4.,
        seed=8888
    )
    
    # Plot 2D frequency-amplitude diagram
    fig1 = plot_frequency_amplitude_diagram(results)
    plt.savefig(f'frequency_amplitude_2D_fE{frac_E:.2f}.png', 
                dpi=150, bbox_inches='tight')
    
    # Plot bifurcation diagrams at selected frequencies
    fig2 = plot_classic_bifurcation_slices(
        results, 
        slice_periods=[1, 3, 5, 10]
    )
    plt.savefig(f'bifurcation_slices_fE{frac_E:.2f}.png', 
                dpi=150, bbox_inches='tight')
    
    # Plot resonance curves at selected amplitudes
    fig3 = plot_amplitude_slices(
        results,
        slice_amplitudes=[0.0, 0.3, 0.6, 0.9]
    )
    plt.savefig(f'resonance_curves_fE{frac_E:.2f}.png', 
                dpi=150, bbox_inches='tight')
    
    plt.show()
    
    # Summary
    print("\n" + "="*70)
    print("ANALYSIS SUMMARY")
    print("="*70)
    print(f"Scanned {len(pulse_amps)} amplitudes × {len(pulse_per_values)} periods")
    print(f"Total simulations: {len(pulse_amps) * len(pulse_per_values)}")
    print(f"Peak threshold: {peak_threshold:.2f}")
    
    # Find regions of maximum response
    max_response_1 = np.max(results['response_std_1'])
    max_idx = np.unravel_index(np.argmax(results['response_std_1']), 
                                results['response_std_1'].shape)
    max_per = pulse_per_values[max_idx[0]]
    max_amp = pulse_amps[max_idx[1]]
    
    print(f"\nOscillator 1 maximum response: {max_response_1:.3f}")
    print(f"  at Period={max_per:.2f} (Freq={1/max_per:.3f}), Amplitude={max_amp:.3f}")
    
    # Propagation failure statistics
    total_failures = np.sum(results['propagation_failure'])
    total_points = results['propagation_failure'].size
    
    print(f"\nPropagation failure detected in {total_failures}/{total_points} parameter combinations")
    print(f"  ({100*total_failures/total_points:.1f}% of parameter space)")
    
    # Peak statistics
    total_peaks_1 = sum(len(p) if p is not None else 0 
                       for row in results['all_peaks_1'] for p in row)
    total_peaks_2 = sum(len(p) if p is not None else 0 
                       for row in results['all_peaks_2'] for p in row)
    print(f"\nTotal suprathreshold peaks (>{peak_threshold}):")
    print(f"  Oscillator 1: {total_peaks_1}")
    print(f"  Oscillator 2: {total_peaks_2}")
```

# 2D Bifurcations

```python
import numpy as np
from scipy.integrate import solve_ivp
import sk_dsp_comm.sigsys as ss
import matplotlib.pyplot as plt

# def sigmoid(u):
#     return np.tanh(u)

# def two_oscillators(t, y, h_ex, h_in, pars, frac_E, frac_I, sr, time_stop, pert):
#     tau_ex, tau_in, c2, c4, c_EE, c_EI = pars
#     time_index = int(t*sr)
#     if time_index >= time_stop*sr:
#         dydt = np.zeros(4)
#         return dydt
#     else:
#         dydt = (
#             (pert[time_index] - y[0] - c2*sigmoid(y[1]) + c_EE*sigmoid(y[0]+frac_E*y[2]))*tau_ex,
#             (h_in             - y[1] - c4*sigmoid(y[1]) + c_EI*sigmoid(y[0]+frac_I*y[2]))*tau_in,
#             (h_ex             - y[2] - c2*sigmoid(y[3]) + c_EE*sigmoid(y[2]+frac_E*y[0]))*tau_ex,
#             (h_in             - y[3] - c4*sigmoid(y[3]) + c_EI*sigmoid(y[2]+frac_I*y[0]))*tau_in
#         )
#     return dydt

# Fixed parameters
h_ex_0 = -7.0
h_in_0 = -4.0
pars = (1, 2.5, 10, 0, 5, 10)  # Homoclinic
frac_E, frac_I = 0.2, 0.

y_ini = [-1.9, -13.56, -1.9, -13.56]

time_stop = 150
sr = 1000
time = np.linspace(start=0, stop=time_stop, num=time_stop*sr)
pulse_wid = 1.

# Parameter ranges for bifurcation diagram
pulse_amp_range = np.linspace(0.6, 2.6, 20)  
pulse_fre_range = np.linspace(0.4, 1.6, 20)  

# Initialize bifurcation diagram matrix
bifurcation_matrix = np.zeros((len(pulse_fre_range), len(pulse_amp_range)))

# Analysis time window (final third: t = 100 to 150)
analysis_start_idx = int((2*time_stop/3) * sr)
analysis_end_idx   = int(   time_stop    * sr)

print("Starting bifurcation analysis...")
print(f"Scanning {len(pulse_amp_range)} x {len(pulse_fre_range)} = {len(pulse_amp_range)*len(pulse_fre_range)} parameter combinations")

# Scan parameter space
for i, pulse_fre in enumerate(pulse_fre_range):
    for j, pulse_amp in enumerate(pulse_amp_range):
        
        # Generate perturbation signal
        pert = h_ex_0 + pulse_amp*ss.rect(
            np.mod(time, 1/pulse_fre) - (1/pulse_fre)/2 + pulse_wid/2, pulse_wid
        )
        
        # Run simulation
        try:
            solution = solve_ivp(
                fun=two_oscillators, 
                t_span=(0, time_stop), 
                y0=y_ini, 
                t_eval=time,
                args=(h_ex_0, h_in_0, pars, frac_E, frac_I, sr, time_stop, pert), 
                method='BDF', 
                max_step=0.1
            )
            
            y_pert = solution.y.T
            
            # Analyze final third (t = 100 to 150)
            y_analysis = y_pert[analysis_start_idx:analysis_end_idx, 2]
            
            # Check if any values exceed 0.5
            if np.any(y_analysis > 0.5):
                bifurcation_matrix[i, j] = 1  # Mark as active
            else:
                bifurcation_matrix[i, j] = 0  # Leave blank (inactive)
                
        except Exception as e:
            print(f"Error at pulse_amp={pulse_amp:.2f}, pulse_per={pulse_fre:.2f}: {e}")
            bifurcation_matrix[i, j] = np.nan
    
    # Progress indicator
    if (i+1) % 5 == 0:
        print(f"Progress: {i+1}/{len(pulse_fre_range)} rows completed")

print("Bifurcation analysis complete!")
```

# Distance Metrics

```python
import numpy as np
from scipy.spatial.distance import euclidean

def path_length(trajectory):
    """Calculate the total path length of a trajectory"""
    displacements = np.linalg.norm(np.diff(trajectory, axis=0), axis=1)
    return np.sum(displacements)

def trajectory_difference_path_based(trajectory_data):
    """
    Compare trajectories based on their actual path characteristics
    """
    osc1_traj = trajectory_data[:, :2]
    osc2_traj = trajectory_data[:, 2:]
    
    # 1. Path length difference
    path1_length = path_length(osc1_traj)
    path2_length = path_length(osc2_traj)
    path_length_diff = abs(path1_length - path2_length)
    
    # 2. Area between paths (integrated distance)
    distances = np.linalg.norm(osc1_traj - osc2_traj, axis=1)
    area_between = np.trapz(distances, dx=1)
    
    # 3. Dynamic differences (velocity/orientation)
    vel1 = np.diff(osc1_traj, axis=0)
    vel2 = np.diff(osc2_traj, axis=0)
    
    # Cosine similarity of velocities (how aligned are movements)
    dot_products = np.sum(vel1 * vel2, axis=1)
    norms_product = np.linalg.norm(vel1, axis=1) * np.linalg.norm(vel2, axis=1)
    cosine_similarities = dot_products / (norms_product + 1e-10)  # avoid division by zero
    avg_alignment = np.mean(cosine_similarities)
    
    return {
        'path_length_difference': path_length_diff,
        'area_between_paths': area_between,
        'velocity_alignment': avg_alignment,
        'composite_measure': path_length_diff + area_between + (1 - avg_alignment)
    }

def dtw_distance(a, b):
    """Simple DTW implementation"""
    n, m = len(a), len(b)
    dtw_matrix = np.zeros((n+1, m+1))
    dtw_matrix[0, 1:] = np.inf
    dtw_matrix[1:, 0] = np.inf
    
    for i in range(1, n+1):
        for j in range(1, m+1):
            cost = abs(a[i-1] - b[j-1])
            dtw_matrix[i, j] = cost + min(dtw_matrix[i-1, j],   # insertion
                                         dtw_matrix[i, j-1],    # deletion  
                                         dtw_matrix[i-1, j-1])  # match
    
    return dtw_matrix[n, m]

def trajectory_difference_dtw(trajectory_data):
    """
    Use DTW to handle trajectories that are similar but phase-shifted
    """
    osc1_traj = trajectory_data[:, :2]
    osc2_traj = trajectory_data[:, 2:]
    
    # Apply DTW to each dimension separately
    dtw_x = dtw_distance(osc1_traj[:, 0], osc2_traj[:, 0])
    dtw_y = dtw_distance(osc1_traj[:, 1], osc2_traj[:, 1])
    
    # Combined DTW distance
    combined_dtw = np.sqrt(dtw_x**2 + dtw_y**2)
    
    return combined_dtw

def trajectory_shape_difference(trajectory_data):
    """
    Compare the shape characteristics of trajectories
    """
    osc1_traj = trajectory_data[:, :2]
    osc2_traj = trajectory_data[:, 2:]
    
    # Calculate curvature for each trajectory
    def calculate_curvature(traj):
        dx_dt = np.gradient(traj[:, 0])
        dy_dt = np.gradient(traj[:, 1])
        d2x_dt2 = np.gradient(dx_dt)
        d2y_dt2 = np.gradient(dy_dt)
        
        curvature = np.abs(dx_dt * d2y_dt2 - dy_dt * d2x_dt2) / (dx_dt**2 + dy_dt**2 + 1e-10)**1.5
        return curvature
    
    curvature1 = calculate_curvature(osc1_traj)
    curvature2 = calculate_curvature(osc2_traj)
    
    # Compare curvature profiles
    curvature_diff = np.mean(np.abs(curvature1 - curvature2))
    
    # Compare turning points (local maxima in curvature)
    from scipy.signal import find_peaks
    peaks1, _ = find_peaks(curvature1, height=np.mean(curvature1))
    peaks2, _ = find_peaks(curvature2, height=np.mean(curvature2))
    
    turning_point_diff = abs(len(peaks1) - len(peaks2))
    
    return curvature_diff + turning_point_diff

def comprehensive_trajectory_analysis(trajectory_data):
    """
    Comprehensive analysis that should distinguish different path types
    """
    osc1_traj = trajectory_data[:, :2]
    osc2_traj = trajectory_data[:, 2:]
    
    results = {}
    
    # Basic point-wise distances (your original approach)
    point_distances = np.linalg.norm(osc1_traj - osc2_traj, axis=1)
    results['max_point_distance'] = np.max(point_distances)
    results['mean_point_distance'] = np.mean(point_distances)
    results['area_between'] = np.trapezoid(point_distances, dx=1)
    
    # Path-based measures
    results['path_length_osc1'] = path_length(osc1_traj)
    results['path_length_osc2'] = path_length(osc2_traj)
    results['path_length_difference'] = abs(results['path_length_osc1'] - results['path_length_osc2'])
    
    # Dynamic measures
    vel1 = np.diff(osc1_traj, axis=0)
    vel2 = np.diff(osc2_traj, axis=0)
    
    velocity_correlations = []
    for i in range(len(vel1)):
        if np.linalg.norm(vel1[i]) > 0 and np.linalg.norm(vel2[i]) > 0:
            cos_sim = np.dot(vel1[i], vel2[i]) / (np.linalg.norm(vel1[i]) * np.linalg.norm(vel2[i]))
            velocity_correlations.append(cos_sim)
    
    results['velocity_correlation'] = np.mean(velocity_correlations) if velocity_correlations else 0
    
    # Shape-based measure
    results['shape_difference'] = trajectory_shape_difference(trajectory_data)
    
    # Create a single composite score
    results['composite_score'] = (
        results['area_between'] + 
        results['path_length_difference'] + 
        (1 - results['velocity_correlation']) +
        results['shape_difference']
    )
    
    return results


result_similar = comprehensive_trajectory_analysis(y_similar[105000:120000, :])

result_different = comprehensive_trajectory_analysis(y_different[105000:120000, :])

result_medium = comprehensive_trajectory_analysis(y_medium[105000:120000, :])

print("Similar trajectories:", result_similar['composite_score'])
print("Different trajectories:", result_different['composite_score'])
print("Medium trajectories:", result_medium['composite_score'])
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

def extract_measures(trajectory_data):
    """
    Extract the three measures from a trajectory
    """
    # Reuse the functions from our previous analysis
    results = comprehensive_trajectory_analysis(trajectory_data)
    
    max_dist = results['max_point_distance']
    path_diff = results['path_length_difference']
    shape_diff = results['shape_difference']
    
    return [max_dist, path_diff, shape_diff]

# Alternative: Individual functions if you want them separate
def max_point_distance(trajectory_data):
    osc1_traj = trajectory_data[:, :2]
    osc2_traj = trajectory_data[:, 2:]
    distances = np.linalg.norm(osc1_traj - osc2_traj, axis=1)
    return np.max(distances)

def path_length_difference(trajectory_data):
    osc1_traj = trajectory_data[:, :2]
    osc2_traj = trajectory_data[:, 2:]
    
    def path_length(traj):
        displacements = np.linalg.norm(np.diff(traj, axis=0), axis=1)
        return np.sum(displacements)
    
    path1 = path_length(osc1_traj)
    path2 = path_length(osc2_traj)
    return abs(path1 - path2)

def shape_difference(trajectory_data):
    osc1_traj = trajectory_data[:, :2]
    osc2_traj = trajectory_data[:, 2:]
    
    def calculate_curvature(traj):
        dx_dt = np.gradient(traj[:, 0])
        dy_dt = np.gradient(traj[:, 1])
        d2x_dt2 = np.gradient(dx_dt)
        d2y_dt2 = np.gradient(dy_dt)
        
        curvature = np.abs(dx_dt * d2y_dt2 - dy_dt * d2x_dt2) / (dx_dt**2 + dy_dt**2 + 1e-10)**1.5
        return curvature
    
    curvature1 = calculate_curvature(osc1_traj)
    curvature2 = calculate_curvature(osc2_traj)
    
    from scipy.signal import find_peaks
    peaks1, _ = find_peaks(curvature1, height=np.mean(curvature1))
    peaks2, _ = find_peaks(curvature2, height=np.mean(curvature2))
    
    curvature_diff = np.mean(np.abs(curvature1 - curvature2))
    turning_point_diff = abs(len(peaks1) - len(peaks2))
    
    return curvature_diff + turning_point_diff

def plot_3d_trajectory_clusters(trajectories, labels=None, colors=None):
    """
    Create a 3D scatter plot of trajectory measures
    
    Parameters:
    trajectories: list of numpy arrays, each shape (samples, 4)
    labels: optional list of labels for each trajectory
    colors: optional list of colors for each trajectory
    """
    fig = plt.figure(figsize=(12, 10))
    ax = fig.add_subplot(111, projection='3d')
    
    # If no colors provided, use a colormap
    if colors is None:
        colors = plt.cm.tab10(np.linspace(0, 1, len(trajectories)))
    elif len(colors) != len(trajectories):
        colors = [colors[0]] * len(trajectories)  # Use first color for all
    
    # Collect all measures for axis limits
    all_measures = []
    
    for index, trajectory in enumerate(trajectories):
        # Calculate the three measures
        measures = [
            max_point_distance(trajectory),
            path_length_difference(trajectory), 
            shape_difference(trajectory)
        ]
        
        all_measures.append(measures)
        
        # Create label
        if labels is not None and index < len(labels):
            label = labels[index]
        else:
            label = f'Traj {index+1}'
        
        # Plot the point
        ax.scatter(*measures, 
                  c=[colors[index]], 
                  s=100, 
                  label=label,
                  alpha=0.7,
                  edgecolors='w',
                  linewidth=0.5)
    
    all_measures = np.array(all_measures)
    
    # Set labels and title
    ax.set_xlabel('Max Point Distance', fontsize=12, labelpad=10)
    ax.set_ylabel('Path Length Difference', fontsize=12, labelpad=10)
    ax.set_zlabel('Shape Difference', fontsize=12, labelpad=10)
    ax.set_title('3D Trajectory Clustering', fontsize=14, fontweight='bold', pad=20)
    
    # Add legend
    ax.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
    
    # Adjust viewing angle for better visualization
    ax.view_init(elev=20, azim=45)
    
    plt.tight_layout()
    plt.show()
    
    return all_measures

# Enhanced version with clustering visualization
def plot_3d_trajectory_clusters_enhanced(trajectories, group_labels=None, cluster_colors=None):
    """
    Enhanced version with better grouping and visualization
    """
    fig = plt.figure(figsize=(14, 10))
    ax = fig.add_subplot(111, projection='3d')
    
    # Define colors for different groups
    if cluster_colors is None:
        cluster_colors = plt.cm.Set1(np.linspace(0, 1, len(set(group_labels)) if group_labels else 1))
    
    all_measures = []
    
    for index, trajectory in enumerate(trajectories):
        measures = [
            max_point_distance(trajectory),
            path_length_difference(trajectory), 
            shape_difference(trajectory)
        ]
        all_measures.append(measures)
        
        # Determine color and label based on grouping
        if group_labels and index < len(group_labels):
            group_idx = group_labels[index]
            color = cluster_colors[group_idx]
            label = f'Group {group_idx}' if index == 0 or group_labels[index-1] != group_idx else ""
        else:
            color = 'blue'
            label = f'Traj {index+1}'
        
        ax.scatter(*measures, 
                  c=[color], 
                  s=100, 
                  label=label,
                  alpha=0.7,
                  edgecolors='w',
                  linewidth=1)
    
    all_measures = np.array(all_measures)
    
    # Set labels with units if known
    ax.set_xlabel('Max Distance\n(larger = more separation)', fontsize=11, labelpad=10)
    ax.set_ylabel('Path Length Diff\n(larger = different path lengths)', fontsize=11, labelpad=10)
    ax.set_zlabel('Shape Difference\n(larger = different shapes)', fontsize=11, labelpad=10)
    ax.set_title('3D Trajectory Measure Space\n(Cluster Analysis)', fontsize=14, fontweight='bold', pad=20)
    
    # Add grid and improve layout
    ax.grid(True, alpha=0.3)
    
    # Remove duplicate labels in legend
    handles, labels = ax.get_legend_handles_labels()
    by_label = dict(zip(labels, handles))
    ax.legend(by_label.values(), by_label.keys(), 
              bbox_to_anchor=(1.15, 1), loc='upper left')
    
    # Multiple viewing angles
    print("Plotting from multiple angles...")
    angles = [(30, 45), (20, 135), (10, 225)]
    
    for i, (elev, azim) in enumerate(angles):
        ax.view_init(elev=elev, azim=azim)
        plt.draw()
        plt.pause(0.5)  # Brief pause to see different angles
    
    plt.tight_layout()
    plt.show()
    
    return all_measures

# Simple version following your pseudocode exactly
def simple_3d_plot(trajectory):
    """
    Simple version following your pseudocode exactly
    """
 
    for index, trajectory in enumerate(trajectories):
        measures = [
            max_point_distance(trajectory),
            path_length_difference(trajectory), 
            max_point_distance(trajectory)
        ]
        
        ax.scatter(*measures, s=80, alpha=0.7, label=f'Traj {index+1}')
    
    ax.set_xlabel('Max Point Distance')
    ax.set_ylabel('Path Length Difference')
    ax.set_zlabel('Shape Difference')
    ax.set_title('3D Trajectory Measures')
    ax.legend()
    
    plt.show()
```

```python
from matplotlib.colors import LinearSegmentedColormap

def create_2d_colormap():
    """Create a colormap that interpolates between four corners"""
    colors = [
        [1, 0, 0],  # Red - bottom-left
        [0, 1, 0],  # Green - bottom-right
        [0, 0, 1],  # Blue - top-left  
        [1, 1, 0],  # Yellow - top-right
    ]
    return LinearSegmentedColormap.from_list('four_corners', colors, N=256)



pulse_per = 4.

# Time array
time_stop = 200
sr        = 1000
time      = linspace(start=0, stop=time_stop, num=time_stop*sr)

rows = time.size


pulse_amps = linspace(0.5, 2, 20)
pulse_wids = linspace(1, 5, 20)


measures_all    = list()
combined_params = list()  # Store combined parameter values for coloring
color_values    = list()
osc_2_amps      = list()

for pulse_wid, pulse_amp in product(pulse_wids, pulse_amps):

    pert = h_ex_0 + pulse_amp*ss.rect(mod(time, pulse_per)-(pulse_per)/2+pulse_wid/2, pulse_wid)

    # Simulation
    y_pert = odeint(func=oscillator, y0=y_ini, t=time, 
                  args=(h_ex_0, h_in_0, pars, frac_E, frac_I, sr, time_stop, pert), 
                  hmax=0.1)

    trajectory = y_pert[100000:130000, :]

    osc_2_extr = trajectory[:, 2].max() - trajectory[:, 2].min()
    osc_2_amps.append(osc_2_extr)
    
    measures = [max_point_distance(trajectory), path_length_difference(trajectory), shape_difference(trajectory)]

    measures_all.   append(measures)
    combined_params.append(pulse_wid * pulse_amp) 

    # Create a combined value that captures both dimensions
    wid_norm = (pulse_wid - pulse_wids.min()) / (pulse_wids.max() - pulse_wids.min())
    amp_norm = (pulse_amp - pulse_amps.min()) / (pulse_amps.max() - pulse_amps.min())
    
    # This formula gives a smooth gradient across the 2D space
    combined_value = (wid_norm + amp_norm) / 2  # or try other combinations
    color_values.append(combined_value)


measures_array = np.array(measures_all)
# combined_params = np.array(combined_params)
color_values   = np.array(color_values)

mask = asarray(osc_2_amps) > 1

#################
# Plot
fig = plt.figure(figsize=(7, 7))
ax  = fig.add_subplot(111, projection='3d')

custom_cmap = create_2d_colormap()

scatter = ax.scatter(measures_array[mask, 0], measures_array[mask, 1], measures_array[mask, 2],
                    c=color_values[mask], cmap=custom_cmap, s=40, alpha=0.7)

cbar = plt.colorbar(scatter, ax=ax, shrink=0.4)
cbar.set_label('Combined Parameter Value')

ax.set_xlabel('Max Distance')
ax.set_ylabel('Path Difference') 
ax.set_zlabel('Shape Difference')
ax.set_title('3 feature space for two-parameter scan - Four Corner Colors');
```

```python
# Initial conditions y_last
y_last = y_ini.copy()

time_stop_pert = 200
sr             = 1000
time           = linspace(start=0, stop=time_stop_pert, num=time_stop_pert*sr)

resolution = 70

pulse_width_min, pulse_width_max = 0.1, 5.1
pulse_ampli_min, pulse_ampli_max = 5, 0.0001

pulse_width = linspace(pulse_width_min, pulse_width_max, resolution)
pulse_ampli = linspace(pulse_ampli_min, pulse_ampli_max, resolution)

pulse_period = 200

results_pert = list()

for pulse_wid, pulse_amp in product(pulse_width, pulse_ampli):

    h_ex_p = h_ex_0 + pulse_amp*ss.rect(mod(time, pulse_period)-pulse_period/2+pulse_wid/2, pulse_wid)

    y = odeint(func=oscillator, y0=y_last, t=time, args=(h_ex_0, h_in_0, pars, frac_E, frac_I, sr, time_stop_pert, h_ex_p),
              hmax=0.1)


    results_pert.append((sum(y[time_stop_pert*sr//2:, 0] - y_last[0])/sr, 
                         sum(y[time_stop_pert*sr//2:, 2] - y_last[2])/sr
                     ))

results_pert_ex1_matrix = asarray(results_pert)[:, 0].reshape(resolution, resolution)
results_pert_ex2_matrix = asarray(results_pert)[:, 1].reshape(resolution, resolution)

print('COMPLETE')
print(' ')

fig, ax = subplots(ncols=2, figsize=(5, 2))

ax[0].hist(results_pert_ex1_matrix.ravel(), bins=20, color='b');

ax[1].hist(results_pert_ex2_matrix.ravel(), bins=20, color='b');

fig.tight_layout()
```

```python
vmin1, vmax1 = 0, 30
vmin2, vmax2 = 0, 30

fig, ax = subplots(figsize=(6, 4), ncols=2)

im1 = ax[0].imshow(results_pert_ex1_matrix.T, cmap='Reds', vmin=vmin1, vmax=vmax1)

ax[0].set_xlabel('Pulse Width', fontsize=8)
ax[0].set_ylabel('Pulse Amplitude', fontsize=8)
ax[0].set_xticks(linspace(0, resolution, 3))
ax[0].set_xticklabels(around(linspace(pulse_width_min, pulse_width_max, 3),1))
ax[0].set_yticks(linspace(0, resolution, 3))
ax[0].set_yticklabels(around(linspace(pulse_ampli_min, pulse_ampli_max, 3), 2))
ax[0].set_title(f'Perturbed. frac_E={frac_E}', fontsize=8)



im2 = ax[1].imshow(results_pert_ex2_matrix.T, cmap='Reds', vmin=vmin2, vmax=vmax2)

ax[1].set_xlabel('Pulse Width', fontsize=8)
# ax[1].set_ylabel('Pulse Amplitude', fontsize=8)
ax[1].set_xticks(linspace(0, resolution, 3))
ax[1].set_xticklabels(around(linspace(pulse_width_min, pulse_width_max, 3), 1))
ax[1].set_yticks(linspace(0, resolution, 3))
ax[1].set_yticklabels(around(linspace(pulse_ampli_min, pulse_ampli_max, 3), 2))


fig.colorbar(im1, ax=ax[0], shrink=0.2);
fig.colorbar(im2, ax=ax[1], shrink=0.2);

fig.tight_layout()

if pars[1] == 1.5:
    
    title_chars = 'Figs/N=1/Fig_2Cb_PulsePert_SNIC.png'
    
elif pars[1] == 2.5:

    title_chars = 'PulsePert_Homoclinic_frac_E=' + str(frac_E) + '_frac_I=' + str(frac_I) +'.png'

# fig.savefig(title_chars, dpi=300, format='png', bbox_inches='tight')

title_chars
```
